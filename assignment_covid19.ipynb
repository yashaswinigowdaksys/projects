```python
from prophet import Prophet
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df_daily = df.groupby('Date')[['Confirmed']].sum().reset_index()
```

```python
df_daily
```

```python
df_prophet = df_daily.rename(columns={'Date': 'ds', 'Confirmed': 'y'})
```

```python
df_prophet
```

```python
df_prophet.columns = ['ds', 'y']
```

```python
df_prophet['ds'] = pd.to_datetime(df_prophet['ds'])
```

```python
model = Prophet(interval_width=0.95)
model.fit(df_prophet)
```

```python
# Predict the next 30 days
future = model.make_future_dataframe(periods=30)
forecast = model.predict(future)
```

```python
future
```

```python
fig = go.Figure()

# Plot Actual Data
fig.add_trace(go.Scatter(x=df_prophet['ds'], y=df_prophet['y'],
                         name='Actual Confirmed', line=dict(color='blue')))
```

```python

# Plot Forecast Trend
fig.add_trace(go.Scatter(x=forecast['ds'], y=forecast['yhat'],
                         name='Predicted Trend', line=dict(color='red')))
```
